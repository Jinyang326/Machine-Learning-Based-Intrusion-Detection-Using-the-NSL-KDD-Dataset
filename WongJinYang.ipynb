```python
# =======================================
#  TASK 2 & TASK 3:
#  DATA PREPROCESSING AND CLASSIFICATION
# =======================================

# Check Missing Values

print("Training Missing Values:",train.isnull().sum().sum())
print("Testing Missing Values:",test.isnull().sum().sum())
```

```python
# Remove Difficulty Column

train = train.drop(columns=['difficulty'])
test = test.drop(columns=['difficulty'])
```

```python
# Attack Mapping

attack_mapping = {

    'normal':'Normal',

    # DoS
    'apache2':'DoS',
    'back':'DoS',
    'land':'DoS',
    'neptune':'DoS',
    'mailbomb':'DoS',
    'pod':'DoS',
    'processtable':'DoS',
    'smurf':'DoS',
    'teardrop':'DoS',
    'udpstorm':'DoS',
    'worm':'DoS',

    # Probe
    'ipsweep':'Probe',
    'mscan':'Probe',
    'nmap':'Probe',
    'portsweep':'Probe',
    'saint':'Probe',
    'satan':'Probe',

    # R2L
    'ftp_write':'R2L',
    'guess_passwd':'R2L',
    'httptunnel':'R2L',
    'imap':'R2L',
    'multihop':'R2L',
    'named':'R2L',
    'phf':'R2L',
    'sendmail':'R2L',
    'snmpgetattack':'R2L',
    'snmpguess':'R2L',
    'spy':'R2L',
    'warezclient':'R2L',
    'warezmaster':'R2L',
    'xlock':'R2L',
    'xsnoop':'R2L',

    # U2R
    'buffer_overflow':'U2R',
    'loadmodule':'U2R',
    'perl':'U2R',
    'ps':'U2R',
    'rootkit':'U2R',
    'sqlattack':'U2R',
    'xterm':'U2R'
}

train['attack_multi'] = train['attack'].map(attack_mapping)
test['attack_multi'] = test['attack'].map(attack_mapping)

train['attack_binary'] = train['attack'].apply(lambda x: 'Normal' if x == 'normal' else 'Attack')
test['attack_binary'] = test['attack'].apply(lambda x: 'Normal' if x == 'normal' else 'Attack')
```

```python
# Split Development Data

train_dev, val_dev = train_test_split(
    train,
    test_size=0.2,
    random_state=42,
    stratify=train['attack_multi']
)
```

```python
# Separate Feature And Labels

X_train = train_dev.drop(
    ['attack', 'attack_binary', 'attack_multi'],
    axis=1
)

X_val = val_dev.drop(
    ['attack', 'attack_binary', 'attack_multi'],
    axis=1
)

X_test = test.drop(
    ['attack', 'attack_binary', 'attack_multi'],
    axis=1
)

y_train_binary = train_dev['attack_binary']
y_val_binary = val_dev['attack_binary']
y_test_binary = test['attack_binary']

y_train_multi = train_dev['attack_multi']
y_val_multi = val_dev['attack_multi']
y_test_multi = test['attack_multi']
```

```python
# Convert labels into machine-readable format

binary_encoder = LabelEncoder()

y_train_binary = binary_encoder.fit_transform(
    y_train_binary
)

y_val_binary = binary_encoder.transform(
    y_val_binary
)

y_test_binary = binary_encoder.transform(
    y_test_binary
)

multi_encoder = LabelEncoder()

y_train_multi = multi_encoder.fit_transform(
    y_train_multi
)

y_val_multi = multi_encoder.transform(
    y_val_multi
)

y_test_multi = multi_encoder.transform(
    y_test_multi
)
```

```python
# Encode Categorical Features

preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        )
    ],
    remainder='passthrough'
)

X_train_encoded = preprocessor.fit_transform(X_train)
X_val_encoded = preprocessor.transform(X_val)
X_test_encoded = preprocessor.transform(X_test)
```

```python
# Scaling 

scaler = StandardScaler(with_mean=False)

X_train_scaled = scaler.fit_transform(X_train_encoded)
X_val_scaled = scaler.transform(X_val_encoded)
X_test_scaled = scaler.transform(X_test_encoded)
```

```python
#======================================
# Task 4: Model Selection and Training
#======================================

# Random Forest - Binary Classification

rf_binary_results = []

for n in [50, 100, 200]:

    rf_binary = RandomForestClassifier(
        n_estimators=n,
        random_state=42,
        n_jobs=-1
    )

    rf_binary.fit(
        X_train_scaled,
        y_train_binary
    )

    val_pred = rf_binary.predict(
        X_val_scaled
    )

    val_acc = accuracy_score(
        y_val_binary,
        val_pred
    )

    rf_binary_results.append(
        [n, val_acc]
    )

    print(
        f"n_estimators={n}, Validation Accuracy={val_acc:.4f}"
    )

# Result Table

rf_binary_table = pd.DataFrame(
    rf_binary_results,
    columns=[
        "n_estimators",
        "Validation Accuracy"
    ]
)

print("\nResult Table: ")
print(rf_binary_table)

# Select Best Parameter

best_n_binary = max(
    rf_binary_results,
    key=lambda x: x[1]
)[0]

print(
    "\nBest n_estimators:",
    best_n_binary
)

# Train Final RF Binary Model

rf_binary_final = RandomForestClassifier(
    n_estimators=best_n_binary,
    random_state=42,
    n_jobs=-1
)

rf_binary_final.fit(
    X_train_scaled,
    y_train_binary
)

# Validation Classification Report

val_pred_binary = rf_binary_final.predict(
    X_val_scaled
)

print("\nClassification report: ")
print(
    classification_report(
        y_val_binary,
        val_pred_binary
    )
)
```

```python
# Random Forest - Multiclass Classification

rf_multi_results = []

for n in [50, 100, 200]:

    rf_multi = RandomForestClassifier(
        n_estimators=n,
        random_state=42,
        n_jobs=-1
    )

    rf_multi.fit(
        X_train_scaled,
        y_train_multi
    )

    val_pred = rf_multi.predict(
        X_val_scaled
    )

    val_acc = accuracy_score(
        y_val_multi,
        val_pred
    )

    rf_multi_results.append(
        [n, val_acc]
    )

    print(
        f"n_estimators={n}, Validation Accuracy={val_acc:.4f}"
    )

# Result Table

rf_multi_table = pd.DataFrame(
    rf_multi_results,
    columns=[
        "n_estimators",
        "Validation Accuracy"
    ]
)

print("\nResult Table: ")
print(rf_multi_table)

# Select Best Parameter

best_n_multi = max(
    rf_multi_results,
    key=lambda x: x[1]
)[0]

print(
    "\nBest n_estimators:",
    best_n_multi
)

# Train Final RF Binary Model

rf_multi_final = RandomForestClassifier(
    n_estimators=best_n_multi,
    random_state=42,
    n_jobs=-1
)

rf_multi_final.fit(
    X_train_scaled,
    y_train_multi
)

# Validation Classification Report

val_pred_multi = rf_multi_final.predict(
    X_val_scaled
)

print("\nClassification report: ")
print(
    classification_report(
        y_val_multi,
        val_pred_multi
    )
)
```

```python
# MLP - Binary Classification

mlp_binary_results = []

for hidden in [

    (50,),
    (100,),
    (100,50)

]:

    mlp_binary = MLPClassifier(
        hidden_layer_sizes=hidden,
        max_iter=30,
        random_state=42
    )

    mlp_binary.fit(
        X_train_scaled,
        y_train_binary
    )

    val_pred = mlp_binary.predict(
        X_val_scaled
    )

    val_acc = accuracy_score(
        y_val_binary,
        val_pred
    )

    mlp_binary_results.append(
        [hidden, val_acc]
    )

    print(
        f"{hidden} -> {val_acc:.4f}"
    )

mlp_binary_table = pd.DataFrame(
    mlp_binary_results,
    columns=[
        "hidden_layer_sizes",
        "Validation Accuracy"
    ]
)

print("\nResult Table: ")
print(mlp_binary_table)

best_hidden_binary = max(
    mlp_binary_results,
    key=lambda x: x[1]
)[0]

print(
    "\nBest Hidden Layer:",
    best_hidden_binary
)

mlp_binary_final = MLPClassifier(
    hidden_layer_sizes=best_hidden_binary,
    max_iter=30,
    random_state=42
)

mlp_binary_final.fit(
    X_train_scaled,
    y_train_binary
)

val_pred_binary = mlp_binary_final.predict(
    X_val_scaled
)

print("\nClassification Report: ")
print(
    classification_report(
        y_val_binary,
        val_pred_binary
    )
)
```
